```python
import yfinance as yf
import matplotlib.pyplot as plt
# import pandas as pd
```

```python
# Start with one liquid instrument. SPY tracks the S&P 500.
# Swap the ticker for anything you like. We are exercising mechanics, not picking a strategy.
data = yf.download("SPY", start="2010-01-01", auto_adjust=True) # The auto_adjust parameter ensures that the data is adjusted for splits and dividends.
```

```python
data
```

```python
prices = data["Close"].squeeze().dropna() # squeeze() handles yfinance's single-column-frame quirk and dropna() removes any missing values.
returns = prices.pct_change().dropna()
```

```python
prices.index
```

```python
returns.index
```

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
# The `figsize` parameter specifies the width and height of the figure in inches (i.e. 11 inches wide and 6 inches tall)
# "nrows" and "ncols" are set to 2 and 1, respectively, to create a 2-row, 1-column layout. "sharex=True" ensures that both subplots share the same x-axis.

ax1.plot(prices.index, prices.values)
# prices.index = x-axis values, e.g. year; prices.values = y-axis values, i.e. price figures

ax1.set_title("SPY price")
ax2.plot(returns.index, returns.values)
ax2.set_title("SPY daily returns")

plt.tight_layout()
# `plt.tight_layout()` automatically adjust spacing between subplots so they don't overlap.
# Without it, titles axis labels, and tick marks from adjacent subplots can overlap each other. It pads the margins and gaps just enough to fit everything neatly.

plt.savefig("step1.png", dpi=120)
# `plt.savefig()` saves the current figure to a file.
# The `fname` parameter specifies the file path/name.
# The `dpi` parameter sets the resolution of the saved image in dots per inch. Higher values yield better quality but larger file sizes.
```
